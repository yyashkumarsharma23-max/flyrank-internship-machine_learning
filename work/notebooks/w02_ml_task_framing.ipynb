## 3. Success metric

**Success Metric: Silhouette Score**

Since this is an unsupervised clustering task, standard metrics like Accuracy do not apply. I will use the Silhouette Score to measure success. This metric evaluates two things: how similar a data point is to its own cluster (cohesion), and how different it is from other clusters (separation).The Silhouette Score ranges from -1 to 1.

What number means 'good'?
A score of 0 means clusters are overlapping, and negative scores mean data points are assigned to the wrong clusters. A score of 0.5 or higher will be considered 'good', as it indicates that the clusters are distinct, well-separated, and meaningful.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
```

A fixed rule relies on rigid thresholds, like if views < 1000 and age > 1 year. However, content performance is multi-dimensional. A post might have low traffic but a very high engagement rate, making it valuable despite failing a simple rule. Writing if/else statements for every possible combination of age, traffic, clicks, and engagement would require hundreds of messy, hard-to-maintain rules. ML (clustering) beats a fixed rule here because it doesn't rely on arbitrary human guesses for thresholds. Instead, it analyzes all features simultaneously across n-dimensions to find the organic, hidden groupings in the data that simple rules would completely miss.
