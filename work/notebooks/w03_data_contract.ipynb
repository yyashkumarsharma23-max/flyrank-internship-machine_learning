## 2. Fields: feature / label / context / excluded

1. **Context (Metadata/Identifiers):**

date, client_hash_id, content_hash_id

Why: Context refers only in the idetification of rows. We cannot use Machine Learning models here.

2. **Label (Target Variable):**

clicks

Why: This refers to a metric in which we can analyze or pedict data outcomes based on previous values.

3. **Features (Inputs):**

impressions: Knowable at the decision moment because Google shows the URL before a user can click on it.

position: Knowable at the decision moment because ranking happens before the user's click action.

4. **Excluded:**

ctr (Click-Through Rate): Excluded because it is directly calculated from clicks (our label). Using it would cause data leakage.

Rows with impressions = 0: Excluded because if a page wasn't seen at all, predicting clicks on it is useless and will skew the data.

```python
query2 = """
SELECT
    report_date, client_hash_id, content_hash_id,  -- Context
    gsc_impressions, gsc_avg_position,             -- Features
    gsc_clicks                                     -- Label
FROM fact_content_daily_performance
WHERE gsc_impressions > 0
LIMIT 5
"""
display(duckdb.sql(query2).df())
```

## 3. Verify it with queries (grain, counts, missing values, windows)

*Every claim above gets a query cell here. A contract claim without a query next to it is a guess.*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
```

**Claim 1: Window**
The data should only contain records for our target window: March 2026. The minimum date should be '2026-03-01' and the maximum date should be '2026-03-31'.

```python
query_window = """
SELECT
    MIN(report_date) AS min_date,
    MAX(report_date) AS max_date
FROM fact_content_daily_performance
WHERE report_date >= '2026-03-01' AND report_date <= '2026-03-31'
"""
display(duckdb.sql(query_window).df())
```

**Claim 2: Missing Values**
There should be zero NULL values in our core context, feature, and label columns within this time window. If gsc_impressions or gsc_clicks are NULL, our ML model will fail to train.

```python
query_missing = """
SELECT
    SUM(CASE WHEN report_date IS NULL THEN 1 ELSE 0 END) AS null_dates,
    SUM(CASE WHEN content_hash_id IS NULL THEN 1 ELSE 0 END) AS null_content,
    SUM(CASE WHEN gsc_impressions IS NULL THEN 1 ELSE 0 END) AS null_impressions,
    SUM(CASE WHEN gsc_clicks IS NULL THEN 1 ELSE 0 END) AS null_clicks
FROM fact_content_daily_performance
WHERE report_date >= '2026-03-01' AND report_date <= '2026-03-31'
"""
display(duckdb.sql(query_missing).df())
```

**Claim 3: Grain**
The grain of this dataset is Daily per Content per Client. This means the combination of (report_date, client_hash_id, content_hash_id) must be strictly unique. There should be no duplicate rows for the same content on the same day.

```python
query_grain = """
SELECT
    report_date,
    client_hash_id,
    content_hash_id,
    COUNT(*) as duplicate_count
FROM fact_content_daily_performance
WHERE report_date >= '2026-03-01' AND report_date <= '2026-03-31'
GROUP BY report_date, client_hash_id, content_hash_id
HAVING COUNT(*) > 1
"""
display(duckdb.sql(query_grain).df())
```

**Claim 4: Counts & Logic**

We need a count of total rows vs. rows valid for training (where gsc_impressions > 0).

Logically, a page cannot have more clicks than impressions. Rows violating this are invalid.

```python
query_counts = """
SELECT
    COUNT(*) AS total_raw_rows,
    SUM(CASE WHEN gsc_impressions > 0 THEN 1 ELSE 0 END) AS valid_training_rows,
    SUM(CASE WHEN gsc_clicks > gsc_impressions THEN 1 ELSE 0 END) AS illogical_rows
FROM fact_content_daily_performance
WHERE report_date >= '2026-03-01' AND report_date <= '2026-03-31'
"""
display(duckdb.sql(query_counts).df())
```

## 4. Data limits

*What can this data never tell you? Unbalanced history, GSC-only early rows, window overlaps.*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
```

**Unbalanced History:** Every client does not have the same amount of data. Some big clients might have thousands of pages and daily records, while smaller clients might only have a few. If we train blindly, the ML model will become biased toward the top 2-3 big clients and perform poorly on smaller ones.

**GSC-only vs GA4 mismatch:** FlyRank combines Google Search Console (GSC) and Google Analytics 4 (GA4). However, not all rows have both. Some clients or older dates might only have GSC data (gsc_data_available = True) but no GA4 data (ga4_data_available = False). We cannot rely on GA4 metrics (like sessions or bounce rate) for all rows.

**Window Overlaps/Sparsity:** We cannot assume every URL gets traffic every single day. There will be gaps (missing days) for specific content_hash_ids if they received 0 impressions, meaning time-series models (like ARIMA or LSTMs) will struggle with broken daily sequences.

```python
# Query 1: Proving Unbalanced History

query_unbalanced = """
SELECT
    client_hash_id,
    COUNT(*) AS total_rows
FROM fact_content_daily_performance
WHERE report_date >= '2026-03-01' AND report_date <= '2026-03-31'
GROUP BY client_hash_id
ORDER BY total_rows DESC
LIMIT 5
"""

# Query 2: Proving GSC-only rows (Missing GA4 data)

query_gsc_ga4 = """
SELECT
    SUM(CASE WHEN gsc_data_available = true THEN 1 ELSE 0 END) AS total_gsc_rows,
    SUM(CASE WHEN ga4_data_available = true THEN 1 ELSE 0 END) AS total_ga4_rows,
    SUM(CASE WHEN gsc_data_available = true AND ga4_data_available = false THEN 1 ELSE 0 END) AS gsc_only_rows
FROM fact_content_daily_performance
WHERE report_date >= '2026-03-01' AND report_date <= '2026-03-31'
"""

print("--- Limitation 1: Unbalanced History (Top 5 Clients dominate the dataset) ---")
display(duckdb.sql(query_unbalanced).df())

print("\n--- Limitation 2: Data Availability Mismatch (GSC vs GA4) ---")
display(duckdb.sql(query_gsc_ga4).df())
```
